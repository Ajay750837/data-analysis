```python
import pandas as pd

# Load dataset
df = pd.read_csv("Employee Data (1).csv")

# Display all rows and columns
with pd.option_context("display.max_rows", None, "display.max_columns", None):
    print(df)

# --------------------------------------------------
# BASIC DATA ANALYSIS
# --------------------------------------------------

# Show first 5 rows
print(df.head())

# Show last 5 rows
print(df.tail())

# Shape of dataset
print("Rows and Columns:", df.shape)

# Column names
print(df.columns)

# Dataset information
print(df.info())

# Statistical summary
print(df.describe())
```

```python
# --------------------------------------------------
# CHECK NULL VALUES
# --------------------------------------------------

print("\nMissing Values:")
print(df.isnull().sum())

# --------------------------------------------------
# FILL MISSING VALUES
# --------------------------------------------------

# Fill numeric columns with mean
numeric_cols = df.select_dtypes(include='number').columns

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].mean())

# Fill text columns with Unknown
text_cols = df.select_dtypes(include='object').columns

for col in text_cols:
    df[col] = df[col].fillna("Unknown")

# --------------------------------------------------
# REMOVE DUPLICATES
# --------------------------------------------------

df.drop_duplicates(inplace=True)

print("\nDuplicates Removed")

# --------------------------------------------------
# STRING OPERATIONS
# --------------------------------------------------

# Convert Name column into string
df["Name"] = df["Name"].astype(str)

# Uppercase
df["Name_Upper"] = df["Name"].str.upper()

# Lowercase
df["Name_Lower"] = df["Name"].str.lower()

# Remove spaces
df["Clean_Name"] = df["Name"].str.strip()

# Name length
df["Name_Length"] = df["Name"].str.len()

# Extract first name
df["First_Name"] = df["Name"].str.split().str[0]

print("\nString Operations Completed")

# --------------------------------------------------
# FILTER DATA
# --------------------------------------------------

# Salary greater than 50000
high_salary = df[df["Salary"] > 50000]

print("\nEmployees With Salary Greater Than 50000:")
print(high_salary)

# IT department employees
it_emp = df[df["Department"] == "IT"]

print("\nIT Department Employees:")
print(it_emp)

# --------------------------------------------------
# SORTING
# --------------------------------------------------

sorted_salary = df.sort_values(by="Salary", ascending=False)

print("\nSorted Salary:")
print(sorted_salary)

# --------------------------------------------------
# GROUP BY
# --------------------------------------------------

# Average salary department wise
avg_salary = df.groupby("Department")["Salary"].mean()

print("\nAverage Salary By Department:")
print(avg_salary)

# Employee count department wise
emp_count = df.groupby("Department")["Employee_ID"].count()

print("\nEmployee Count By Department:")
print(emp_count)

# --------------------------------------------------
# PIVOT TABLE
# --------------------------------------------------

pivot = pd.pivot_table(
    df,
    values="Salary",
    index="Department",
    aggfunc=["mean", "sum", "count"]
)

print("\nPivot Table:")
print(pivot)

# --------------------------------------------------
# VISUALIZATION
# --------------------------------------------------

import matplotlib.pyplot as plt

# Bar Chart
avg_salary.plot(kind='bar')

plt.title("Average Salary By Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")

plt.show()

# Histogram
df["Salary"].plot(kind='hist', bins=10)

plt.title("Salary Distribution")
plt.xlabel("Salary")

plt.show()

# --------------------------------------------------
# SAVE CLEANED DATA
# --------------------------------------------------

df.to_csv("Cleaned_Employee_Data.csv", index=False)

print("\nCSV File Saved")

# Save Excel file
df.to_excel("Cleaned_Employee_Data.xlsx", index=False)

print("Excel File Saved")

# --------------------------------------------------
# FINAL MESSAGE
# --------------------------------------------------

print("\nData Analysis Completed Successfully")
```
