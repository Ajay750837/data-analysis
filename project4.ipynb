```python
gender_count = df["Gender"].value_counts()
gender_count.plot(kind="bar", color=['skyblue', 'lightcoral'])
plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
travel_count = df["Type of Travel"].value_counts()
travel_count.plot(kind="bar", color='lightgreen')
plt.title("Types of Travel Distribution")
plt.xlabel("Types of Travel")
plt.ylabel("Number of Passengers")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns 
table = pd.crosstab(df["Class"],
                    df["satisfaction"])
sns.heatmap(table,
            annot=True,
            fmt="d",
            cmap="Blues")
plt.title("Class vs satisfaction")
plt.show()
```

```python
num = df.select_dtypes(include=["int64","float64"])
display(num)
```

```python
corr = num.corr()
corr.round(2)
```
